## Using postgres via Docker
```bash
docker run -it --rm `
  -e POSTGRES_USER="root" `
  -e POSTGRES_PASSWORD="root" `
  -e POSTGRES_DB="ny_taxi" `
  -v ny_taxi_postgres_data:/var/lib/postgresql/data `
  -p 5433:5432 `
  postgres:13
  ```

## Check database
```bash
 uv run pgcli -h localhost -p 5433 -u root -d ny_taxi
```

## Ingest data
```bash
uv run python ingest_data.py `
  --pg-user="root" `
  --pg-pass="root" `
  --pg-host="localhost" `
  --pg-port=5433 `
  --pg-db="ny_taxi" `
  --target-table="yellow_taxi_trips" `
  --chunksize=50000
  ```

## Pull pgadmin container:
```bash
docker run -it `
  -e PGADMIN_DEFAULT_EMAIL="admin@admin.com" `
  -e PGADMIN_DEFAULT_PASSWORD="root" `
  -v pgadmin_data:/var/lib/pgadmin `
  -p 8085:80 `
  dpage/pgadmin4
```

## Create docker network => pgadmin vs postgresql container is the same network
```bash
docker network create pg-network
```

Check
```bash
docker network ls
```

## run in two different containers:
### 1st: SQL container

```bash
docker run -it `
  -e POSTGRES_USER="root" `
  -e POSTGRES_PASSWORD="root" `
  -e POSTGRES_DB="ny_taxi" `
  -v ny_taxi_postgres_data:/var/lib/postgresql `
  -p 5433:5432 `
  --network=pg-network `
  --name pgdatabase `
  postgres:18
```
### 2nd: pgadmin with network config
```bash
docker run -it `
  -e PGADMIN_DEFAULT_EMAIL="admin@admin.com" `
  -e PGADMIN_DEFAULT_PASSWORD="root" `
  -v pgadmin_data:/var/lib/pgadmin `
  -p 8085:80 `
  --network=pg-network `
  --name pgadmin `
  dpage/pgadmin4
```

Open Port to access pgadmin UI
```bash
Host name: pgdatabase (Docker container name for PostgreSQL)
Port: 5432 (internal port exposed by PostgreSQL container)
Username: root
Password: root
```
