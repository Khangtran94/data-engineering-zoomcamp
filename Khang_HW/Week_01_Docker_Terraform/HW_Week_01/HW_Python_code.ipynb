```python
import pandas as pd
# Download and save the Parquet file
parquet_url = "https://d37ci6vzurychx.cloudfront.net/trip-data/green_tripdata_2025-11.parquet"
df_green = pd.read_parquet(parquet_url)
df_green.to_parquet('green_tripdata_2025-11.parquet')

# Download and save the CSV file
csv_url = "https://github.com/DataTalksClub/nyc-tlc-data/releases/download/misc/taxi_zone_lookup.csv"
df_zones = pd.read_csv(csv_url)
df_zones.to_csv('taxi_zone_lookup.csv', index=False)

print("Files downloaded successfully!")
```

## Question 5. Biggest pickup zone

Which was the pickup zone with the largest `total_amount` (sum of all trips) on November 18th, 2025?

- East Harlem North
- East Harlem South
- Morningside Heights
- Forest Hills

```python
november_18 = df_green[df_green['lpep_pickup_datetime'].dt.strftime('%Y-%m-%d') == '2025-11-18'][['PULocationID','total_amount']]
zone = november_18.merge(df_zones[['LocationID', 'Zone']],left_on = 'PULocationID',right_on ='LocationID')
zone
```

```python
zone.groupby('Zone')['total_amount'].sum().to_frame('Total_amount').reset_index().sort_values('Total_amount',ascending=False).head()
```
